```python
import pandas as pd
import numpy as np

project_path = r"C:\Users\Bawan Singh\OneDrive\Documents\Arshleen\market-portfolio-risk-analysis"

returns = pd.read_csv(
    project_path + r"\data\processed\daily_returns.csv",
    index_col=0,
    parse_dates=True
)

portfolio_returns = pd.read_csv(
    project_path + r"\data\processed\portfolio_returns.csv",
    index_col=0,
    parse_dates=True
)
```

```python
portfolio_annualized_return = pd.read_csv(
    project_path + r"\data\processed\portfolio_annualized_return.csv",
    index_col=0
).squeeze()
```

```python
portfolio_annualized_volatility = pd.read_csv(
    project_path + r"\data\processed\portfolio_annualized_volatility.csv",
    index_col=0
).squeeze()
```

```python
sharpe_ratio = pd.read_csv(
    project_path + r"\data\processed\sharpe_ratio.csv",
    index_col=0
).squeeze()
```

```python
benchmark_returns=returns["SPY"]
```

```python
portfolio_beta = {}

for portfolio in portfolio_returns.columns:
    covariance = portfolio_returns[portfolio].cov(benchmark_returns)
    benchmark_variance = benchmark_returns.var()
    
    portfolio_beta[portfolio] = covariance / benchmark_variance

portfolio_beta
```

```python
portfolio_beta_table = pd.Series(portfolio_beta)

portfolio_beta_table
```

```python
treynor_ratio = portfolio_annualized_return / portfolio_beta_table

treynor_ratio
```

```python
portfolio_beta_table
```

```python
portfolio_annualized_return
```

```python
treynor_ratio_table=pd.DataFrame({ "Treynor Ratio":treynor_ratio})
treynor_ratio_table.style.format("{:.2f}")
```

```python
benchmark_annualized_return = benchmark_returns.mean() * 252

benchmark_annualized_return
```

```python
jensens_alpha = (
    portfolio_annualized_return
    - (portfolio_beta_table * benchmark_annualized_return)
)

jensens_alpha
```

```python
jensens_alpha_table = pd.DataFrame({
    "Jensen's Alpha": jensens_alpha
})

jensens_alpha_table.style.format("{:.2%}")
```

```python
jensens_alpha.to_csv(
    project_path + r"\data\processed\jensens_alpha.csv"
)
```
