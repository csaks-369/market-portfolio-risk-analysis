```python
final_dashboard = pd.read_csv(
    project_path + r"\data\processed\final_dashboard.csv",
    index_col=0
)

final_dashboard
```

```python
# Display the key portfolio results
summary_columns = [
    "Annualized Return",
    "Annualized Volatility",
    "Sharpe Ratio",
    "Sortino Ratio",
    "Beta",
    "Maximum Drawdown"
]

executive_summary = final_dashboard[summary_columns].copy()

executive_summary
```

```python
# Compare the portfolios using the key metrics

for portfolio in executive_summary.index:
    print(f"\n{portfolio} Portfolio")
    print("-" * 30)

    print(
        f"Annualized Return: "
        f"{executive_summary.loc[portfolio, 'Annualized Return']:.2%}"
    )

    print(
        f"Annualized Volatility: "
        f"{executive_summary.loc[portfolio, 'Annualized Volatility']:.2%}"
    )

    print(
        f"Sharpe Ratio: "
        f"{executive_summary.loc[portfolio, 'Sharpe Ratio']:.3f}"
    )

    print(
        f"Sortino Ratio: "
        f"{executive_summary.loc[portfolio, 'Sortino Ratio']:.3f}"
    )

    print(
        f"Beta: "
        f"{executive_summary.loc[portfolio, 'Beta']:.3f}"
    )

    print(
        f"Maximum Drawdown: "
        f"{executive_summary.loc[portfolio, 'Maximum Drawdown']:.2%}"
    )
```

```python
print("EXECUTIVE ANALYSIS")
print("=" * 60)

print("""
The analysis shows a clear trade-off between return generation and portfolio risk.

The Growth portfolio produced the highest annualized return at 18.73%.
However, it also had the highest annualized volatility at 19.60%,
the highest beta at 0.962, and the largest maximum drawdown at -30.31%.

The Conservative portfolio had the lowest annualized volatility at 10.81%
and the lowest beta at 0.369, indicating lower sensitivity to movements
in the S&P 500 than the other portfolios.

The Balanced portfolio generated a 12.32% annualized return with 12.60%
annualized volatility and a beta of 0.573.

Interestingly, the Balanced portfolio had the highest Sharpe Ratio at
approximately 0.978, while the Conservative portfolio had the highest
Sortino Ratio at approximately 1.587.

This demonstrates that the portfolio with the highest absolute return
is not necessarily the portfolio with the strongest risk-adjusted
performance.

The results highlight the importance of evaluating portfolio performance
using multiple measures rather than relying on return alone.
""")
```

```python
# Risk analysis

risk_columns = [
    "Historical VaR (95%)",
    "Expected Shortfall (95%)",
    "Maximum Drawdown",
    "Maximum Rolling Volatility"
]

risk_analysis = final_dashboard[risk_columns].copy()

risk_analysis
```

```python
# Stress-period analysis

stress_columns = [
    "2020 Stress Period Return",
    "2022 Annual Return"
]

stress_analysis = final_dashboard[stress_columns].copy()

stress_analysis
```

```python
print("STRESS PERIOD ANALYSIS")
print("=" * 60)

print("""
The selected stress-period analysis shows that portfolio behavior
varied across different market environments.

During the February-June 2020 stress window, all three portfolios
generated positive cumulative returns over the selected period.

In contrast, during calendar year 2022, all three portfolios experienced
negative returns, with the Growth portfolio experiencing the largest
decline at -26.54%.

The 2022 results illustrate the impact that a broad risk-off environment
can have on portfolios with greater exposure to equity and growth assets.

These stress-period results should be interpreted alongside the VaR,
Expected Shortfall, volatility and maximum drawdown measures rather
than used as standalone measures of future risk.
""")
```
