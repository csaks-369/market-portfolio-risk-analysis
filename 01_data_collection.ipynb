```python
import yfinance as yf
import pandas as pd
import numpy as np

tickers = ["SPY", "QQQ", "TLT", "GLD"]

market_data = yf.download(
    tickers=tickers,
    start="2019-01-01",
    end="2026-01-01",
    auto_adjust=True,
    progress=True
)

market_data.head()
```

```python
market_data.tail()
```

```python
market_data.info()
```

```python
prices = market_data["Close"]
prices.head()
```

```python
march_13 = prices.loc["2020-03-13", "SPY"]
march_16 = prices.loc["2020-03-16", "SPY"]

return_calculation = (march_16 - march_13) / march_13

print(return_calculation)
```

```python
returns = prices.pct_change()
returns.head()
```

```python
returns = returns.dropna()
returns.head()
```

```python
correlation = returns.corr()
correlation
```

```python
returns.std()
```

```python
returns.mean()
```

```python
summary=pd.DataFrame({"Average Daily Return":returns.mean(),"Daily Volatility":returns.std()}) 
summary
```

```python
summary_percent = summary * 100
summary_percent
```

```python
annualized_return = returns.mean() * 252

annualized_return
```

```python
annualized_volatility = returns.std() * np.sqrt(252)

annualized_volatility
```

```python
summary = pd.DataFrame({
    "Annualized Return": annualized_return,
    "Annualized Volatility": annualized_volatility
})

summary
```
